```python
# series- 1 dimensional labled array
s=pd.Series([1,2,3,4,5])
print(s)
print(type(s))

# index 
print(s[0])
print(s[2])
```

```python
# index and labled are different means index tell the position of th element 
# but labeled use is named index, column that are print in the output before 
# the real value is labled that are by default == index value
s=pd.Series([23,4,5,67], index=["Adam", "Eve", "Bob", "Charlie"])
print(s)
# here we change the labled name
# and now we can access our data using lable value and index value both 
print(s["Adam"])
# print(s[0])
print(s.index)
```

```python
# properties
import pandas as pd
# homogeneous
s=pd.Series([23,4,5,"abc"], index=["Adam", "Eve", "Bob",12])
print(s)# datatypes of all value are same 
# we can perform vectorization
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])
print(s1+s2)
# mutable value
s1[0]=100
print(s1)
# immutable size
changed_s1=s1.drop(2)
print(s1)# size of s1 not affected value change but in changed_s1 means if we want to add or drop an element than we need to creat another series 
print(changed_s1)
# handle missing value with NaN
```

```python
# second data structure 
# data frame - Table form
# many ways to create dataFrame
# 1st way
info={
    "Name":["Adam","Eve","Bob","jack"],
    "CGPA":[9.3, 8.4, 6.4, 8.5],
    "Age":[23, 24, 19, 20]
}
df=pd.DataFrame(info)
print(df, type(df))
print(df.index)# print the lable for row value
print(df.columns)# print the lable for columns
df
```

```python
# 2nd way
df=pd.DataFrame([["adam",23],["eve",24],["bob",19],["jack",20]], columns=["name", "age"])
print(df)
df
```

```python
# 3rd ways
import numpy as np
np_arr=np.array([[1, 2, 3], [4, 5, 6]])
df=pd.DataFrame(np_arr, columns=["A","B","C"])
print(df)
```

```python
# Accessing rows value
# label base access
df.loc[2]
df.loc[0:2] # start idx : end idx (inclusive)
# index base
df.iloc[2]
df.iloc[0:2] #start idx : end idx (exclusive)

# Accessing row and column together
# for label base
df.loc[0:2, ["city", "aqi"]]
df.loc[0:2, ["country", "city", "latitude", "longitude"]]
df.iloc[0:3,2]# "2" is index of city
df.iloc[0:3, 1:5]# row index range, column index range
# cell
# selecting single numerical value
df.at[0, "city"]
df.iat[0, 2]
# pandas provide view of data like numpy
# means
cities=df["city"]
cities[0]="New york"
# to avoid this mistake
cities=df["city"].copy()
cities[0]="Mumbai"# copy
cities
# now when we print cities value its change but in orginal data the value of city remains same
```

```python
df
```

```python
# filtering
# single condition
df[ df["aqi"] > 100]

# Multiple condition - use &(and) |(or)  ^(not)
df[ (df["aqi"] > 100) & (df["temperature"] < 30)]
df[ (df["aqi"] > 100) & (df["temperature"] > 30)]
# simple pattern - when temperature is low aqi is low in maximum city
# if we want to print only limited columm
aqi_data=df[ (df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi", "temperature"]]
# pandas not change label value after filtering

aqi_data.iloc[0]
# aqi_data.loc[0] # show an error because label 0 are not exist 
```

```python
# Query method for data filtering 
# rule for query()
# expression as a string
# column based reference
# backticks for column_name with space or special chars
# operators ( &, |, ~, >=, >, ==, etc.)
df.query("aqi > 100")
# @ to reference python variable 
aqi_var=100
df.query("aqi > @aqi_var & temperature > 30")[["city", "aqi"]]
# chain comparison
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
"""Note - query returns copy of filter data not view"""
# dont nedd to remeber all this copy and view things for the shake of simplicity 
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]].copy()
```

```python
"""Data cleaning"""
import pandas as pd
# handling missing value
df=pd.read_csv("raw_data.csv")
df.isnull() # detect NaN value
df.isna() # both are same
df.isnull().sum() # number of missing value inn every column
df.dropna() # drop all the row where any nan value exist 
df.dropna(axis=1) # drop all the column where any nan value exist 
df.fillna(0) # all NaN value will repalace by 0 
df.fillna(100) # all NaN value will replace by 100
age_mean=df["age"].mean()
cleaned_data = df.copy()
age_mean=cleaned_data["age"].mean()
cleaned_data["age"]=cleaned_data["age"].fillna(age_mean)
cleaned_data
# ffill() stands for forward fill
df.ffill() # all nan value replace by the value before the nan value
df.bfill() # all nan value replace by the value after the nan value
```

```python
df
```

```python
# handle duplicate value
df.duplicated()
# can check duplicate value for column wise, row wise, and set of column and row wise
df["country"].duplicated()
df[["country", "gender"]].duplicated()
df.drop_duplicates() # drop all the duplicate value
"""Note:- this changes not reflects in original data to do this we need to copy dat and then proceed or 
we can use "implace=True" attribute"""
df2=df.copy() # prefered may be we need some times to use our original data show if we have copy of it then its simple for us
df.drop_duplicates(inplace=True) # chnnges show in the original data
```
