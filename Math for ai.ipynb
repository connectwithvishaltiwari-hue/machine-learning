```python
# binomial ditribution

from scipy.stats import binom

n = 5
k = 3
p = 0.5

prob = binom.pmf(k, n, p)
print(prob) # 5/16 = 0.3124
```

```python
# Uniform distribution

import numpy as np
import matplotlib.pyplot as plt

# Values from 0 to 10
values = np.random.uniform(0, 10, 10_00_000)
#                         (start , end , number of values)  Start and end is basically a range 
# ploting a histogram - becomes more uniform with more values
plt.hist(values , bins=100, density = True, alpha = 0.3)

plt.title("continuous uniform distribution between 0 to 10")
plt.xlabel("x")
plt.ylabel("Probabililty density")
plt.grid(True)
plt.show()
```

```python
# Normal distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# parameters 
mu = 70 # mean
sigma = 10 # standard deviation
# X-axis Values
x = np.linspace (30, 110, 1000)

# normal PDF
y = norm.pdf(x, mu, sigma)

# plots the bell curve
plt.plot(x, y, color = "black", linewidth = 2, label = "Normal Distribution")

# Shade 68% region ( mu +- 1sigma)
x1=np.linspace(mu-sigma, mu+sigma, 1000)
plt.fill_between(x1,norm.pdf(x1, mu, sigma), color="green", alpha=0.6, label="68% region (1sigma)")

# Shade 95% region ( mu +- 2sigma)
x2=np.linspace(mu-2*sigma, mu+2*sigma, 1000)
plt.fill_between(x2, norm.pdf(x2, mu, sigma), color="yellow", alpha=0.5, label="95% region (2sigma)")

# Shade 99.7% region ( mu +- 3sigma)
x3=np.linspace(mu-3*sigma, mu+3*sigma, 1000)
plt.fill_between(x3,norm.pdf(x3, mu, sigma), color="blue", alpha=0.4, label="99.7% region (3sigma)")


# labels and title
plt.title("Normal distribution of student scores")
plt.xlabel("Score")
plt.ylabel("Probability Density")
plt.legend()
plt.show()

```

```python
# CAlCULAS
import numpy as np
import matplotlib.pyplot as plt

# Define X values (X >= 0 for sqrt)
x = np.linspace(0, 100, 100_000)
y = np.sqrt(x)

# x plot
plt.plot(x, y, label = "y = Root x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = root x")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# graph of sin x
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2*np.pi, 2*np.pi, 100)
y = np.sin(x)

# Plot the graph
plt.plot(x, y, label = 'y = sin(x)', color='blue')

# highlights x=0 and y=0 lines
plt.axvline(0, color="black", linewidth=1.5, linestyle='--', label='x=0')
plt.axhline(0, color="green", linewidth=1.5, linestyle='--', label='y=0')

plt.xlabel('x')
plt.xlabel('y')

plt.title("Graph of y = sin(x)")
plt.grid(True)
plt.legend()
plt.show()
```
